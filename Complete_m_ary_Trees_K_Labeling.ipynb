```python
"""
# below is code for an alternate but still effective traversing method
# level order for each child of main root uses for loops and not while loops
# only the three following functions are different

# creates children for all children of the main root
# except the last main child of the main root and the children
# of those children down
def second_layer_down(root,h,m,k,array,depth):
      # if maximum depth is reached the function returns
      if depth == h:
        return
      #assigning children to root in argument
      for i in range(0,m):
            e = array[-1]
            while e in array:
              e = e + 1
            array.append(e)
            child_node = Node(e-root.label,e)
            root.add_child(child_node)
      # through recursion adding children to each child just created
      for child in root.children:
              second_layer_down(child, h, m, array, depth + 1)

  #creates children of the mth (last) child of main root
  #and the children of those children down
def mth_subtree(root,h,m,array,V,depth):
      # if maximum depth is reached the function returns
      if depth == h:
        return
      # assigning children to root in argument
      for i in range(0,m):
            while V in array:
              V = V - 1
            child_node = Node(V-root.label,V)
            root.add_child(child_node)
            array.append(V)
      # through recursion adding children to each child just created
      for child in root.children:
          mth_subtree(child, h, m, array, V, depth + 1)

  # Create the tree
def m_aray_tree(m,h):
    # assigning initial node (root)
    root = Node(1,0)
    # assigning first child of root
    child1 = Node(0,0)
    root.add_child(child1)
    #V is the number of vertices and the
    #largest edge value should equal this. The number of edges is V-1.
    #The largest value of a label for any vertices should be math.ceil(V/2)
    V = (m ** (h + 1) - 1) / (m - 1)
    d = math.ceil(V/2) / (m -1)
    # l is label
    l = 0
    edges = []
    for i in range(1,m):
      l = l + d
      lnew = math.floor(l)
      #in first layer each edge is the label of the child node plus
      #the value of the root (1)
      edge = lnew + 1
      # every time an edge will be assigned to a node it will be added to edges array
      edges.append(edge)
      child = Node(lnew,edge)
      root.add_child(child)
    child1.label = 1
    child1.edge = 2
    edges.append(2)
    for child in root.children:
      # for children of root 1 to m-1
      if child != root.children[-1]:
        second_layer_down(child,h,m,edges,depth=1)
      # for last child of root
      if child == root.children[-1]:
        mth_subtree(child,h,m,edges,int(V),depth=1)
    return root

"""



import math

# creates node class and the attributes of each node
# the edge of a node is the edge connecting a node to its root
class Node:
    def __init__(self, label,edge):
        self.label = label
        self.edge = edge
        self.children = []

    def add_child(self, child):
        self.children.append(child)

# function to print tree
def print_tree(node, level=0):
    if node.edge == 0:
      print('    ' * level + str(node.label))
    else:
      print('    ' * level + str(node.edge) + '-' + str(node.label))
    for child in node.children:
        print_tree(child, level + 1)

"""creates children for all children of the main root
except the last main child of the main root and the children
of those children down"""
def second_layer_down(root,h,m,edgearray,depth):
    # if maximum depth is reached the function returns
    if depth == h:
      return
    k = 0
    # k is how many children have been created or added to a parent so far
    while k < m:
      # determining next value for edge labeling
      e = edgearray[-1]
      # if edge previously used add one until not used befroe
      while e in edgearray:
            e = e + 1
      child_node = Node(e-root.label,e)
      root.add_child(child_node)
      edgearray.append(e)
      # through recursion adding children and labeling
      second_layer_down(root.children[k], h, m, edgearray, depth + 1)
      k = k + 1

"""creates children of the mth (last) child of main root
and the children of those children down"""
def mth_subtree(root,h,m,edgearray,V,depth):
    # if maximum depth is reached the function returns
    if depth == h:
      return
    # k is how many children have been created or added to a parent so far
    k = 0
    while k < m:
      # determining next value for edge labeling
      while V in edgearray:
            V = V - 1
      child_node = Node(V-root.label,V)
      root.add_child(child_node)
      edgearray.append(V)
      # through recursion adding children and labeling
      mth_subtree(root.children[k], h, m, edgearray, V, depth + 1)
      k = k + 1

# Create the tree
def m_aray_tree(m,h):
  # assigning initial node (root)
  root = Node(1,0)
  # assigning first child of root
  child1 = Node(0,0)
  root.add_child(child1)
  """ V is the number of vertices and the largest edge value should equal this.
  The number of edges is V-1. The largest value of a label for any vertices
  should be math.ceil(V/2)"""
  V = (m ** (h + 1) - 1) / (m - 1)
  d = math.ceil(V/2) / (m -1)
  # l is label
  l = 0
  edges = []
  for i in range(1,m):
    l = l + d
    lnew = math.floor(l)
    """in first layer each edge is the label of the child node plus
    the value of the root"""
    edge = lnew + 1 # value of root is 1
    # every time an edge will be assigned to a node it will be added to edges array
    edges.append(edge)
    child = Node(lnew,edge)
    root.add_child(child)
  child1.label = 1
  child1.edge = 2
  edges.append(2)
  for child in root.children:
    # for children of root 1 to m-1
    if child != root.children[-1]:
      second_layer_down(child,h,m,edges,depth=1)
    # for last child of root
    if child == root.children[-1]:
      mth_subtree(child,h,m,edges,int(V),depth=1)
  return root

# Print the tree
print_tree(m_aray_tree(5,3))
```
